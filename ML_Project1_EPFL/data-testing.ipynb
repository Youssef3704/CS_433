```python
import pandas as pd
```

```python
x_train = pd.read_csv('data/81fdc2af-6811-4089-9054-9d59f417f819_dataset/dataset/x_train.csv', sep=',')
y_train = pd.read_csv('data/81fdc2af-6811-4089-9054-9d59f417f819_dataset/dataset/y_train.csv', sep=',')
x_train.head()

```

```python
y_train.head()
```

```python
df = x_train.merge(y_train, on="Id", how="inner")
df.head()
```

```python
df.columns
```

```python
for c in df.columns:
    print(c)

print(len(df.columns))
```

```python
label = "_MICHD"
df["target"] = (df[label] == 1).astype(int)
corr = (
    df.
    drop(columns=[label, "Id"], errors="ignore")
    .corr()["target"].drop("target")
)
```

```python
for name, value in corr.abs().sort_values(ascending=False).head(100).items():
    print(f"{name:12s} {value:.3f}")
```
